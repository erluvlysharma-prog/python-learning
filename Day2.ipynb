```python
# 1.  Check whether a number is positive, negative, or zero.
n = int(input("Enter a number: "))
if n > 0:
  print("Number is positive")
elif n < 0:
  print("Number is negative")
else:
  print("Number is zero")
```

```python
 # 2. Check whether a number is even or odd.
n = int(input("Enter a number: "))
if n%2 == 0:
  print("Number is even")
else:
  print("Number is odd")
```

```python
# 3.Find the larger of two numbers.
a = int(input("Enter the first number: "))
b = int(input("Enter the first number: "))
if a > b:
  print("First number is larger")
elif b > a:
  print("Second number is larger")
else:
  print("Both numbers are equal")
```

```python
# 4.  Find the largest of three numbers.
a = int(input("Enter the first number: "))
b = int(input("Enter the first number: "))
c = int(input("Enter the first number: "))
if a>b and a>c:
  print("Frist number is largest")
elif b>a and b>c:
  print("Second number is largest")
else :
  print("Third number is largest")
```

```python
# 6.  Check whether a person is eligible to vote based on age.
age = int(input("Enter the age: "))
if age >= 18:
  print("Person is eligible to vote")
else:
  print("Person is not eligible to vote")
```

```python
# 7.  Check whether a year is a leap year.
year = int(input("Enter the year:"))
if year % 4 == 0:
  print("Year is a leap year")
else:
  print("Year is not a leap year")
```

```python
# 8.  Check whether a character is a vowel or consonant.
char = input("Enter the character: ")
if char == 'a' and char == 'e' and char == 'i' and char == 'o' and char == 'u':
  print("Character is a vowel")
else:
  print("Character is a consonant")
```

```python
# 9.  Check whether a character is an alphabet, digit, or special character.
char = input("Enter the character: ")
if char.isalpha():
  print("Character is an alphabet")
elif char.isdigit():
  print("Character is a digit")
else:
  print("Character is a special character")
```
