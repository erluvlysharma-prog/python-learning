```python
# 26. Convert temperature between Celsius, Fahrenheit, and Kelvin using a menu.

print("1. Celsius to Fahrenheit")
print("2. Fahrenheit to Celsius")
print("3. Celsius to Kelvin")
print("4. Kelvin to Celsius")
print("5. Fahrenheit to Kelvin")
print("6. Kelvin to Fahrenheit")

choice = int(input("Enter your choice: "))

temp = float(input("Enter temperature: "))

match choice:
    case 1:
        print("Fahrenheit: ", (temp * 9/5) + 32)
    case 2:
        print("Celsius: ", (temp - 32) * 5/9)
    case 3:
        print("Kelvin: ", temp + 273.15)
    case 4:
        print("Celsius: ", temp - 273.15)
    case 5:
        print("Kelvin: ", (temp - 32) * 5/9 + 273.15)
    case 6:
        print("Fahrenheit: ", (temp - 273.15) * 9/5 + 32)
    case _:
        print("Invalid choice")

```

```python
# 29. Create a menu for checking even/odd, positive/negative, or prime

print("1. Check Even/Odd")
print("2. Check Positive/Negative")
print("3. Check Prime")

choice = int(input("Enter your choice (1-3): "))
num = int(input("Enter a number: "))

match choice:
    case 1:
        if num % 2 == 0:
            print("Even")
        else:
            print("Odd")

    case 2:
        if num > 0:
            print("Positive")
        elif num < 0:
            print("Negative")
        else:
            print("Zero")

    case 3:
        if num < 2:
            print("Not Prime")
        else:
            is_prime = True

            for i in range(2, num):
                if num % i == 0:
                    is_prime = False
                    break

            if is_prime:
                print("Prime")
            else:
                print("Not Prime")

    case _:
        print("Invalid choice")
```

```python
# 30. Create a simple ATM menu: balance, deposit, withdraw, exit.

print("1. Check Balance")
print("2. Deposit")
print("3. Withdraw")
print("4. Exit")

choice = int(input("Enter your choice (1-4): "))
balance = 1000

match choice:
    case 1:
        print("Balance:", balance)
    case 2:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Balance:", balance)
    case 3:
        amount = float(input("Enter withdrawal amount: "))
        if amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Balance:", balance)
    case 4:
        print("Thank you for using the ATM")
    case _:
        print("Invalid choice")
```

```python
# 32. Create a menu-driven number utility:
# factorial, reverse, digit sum, palindrome

print("1. Factorial")
print("2. Reverse")
print("3. Digit Sum")
print("4. Palindrome")

choice = int(input("Enter your choice (1-4): "))
num = int(input("Enter a number: "))

match choice:

    case 1:
        factorial = 1

        for i in range(1, num + 1):
            factorial = factorial * i

        print("Factorial:", factorial)

    case 2:
        reverse = 0
        temp = num

        while temp > 0:
            digit = temp % 10
            reverse = reverse * 10 + digit
            temp = temp // 10

        print("Reverse:", reverse)

    case 3:
        digit_sum = 0
        temp = num

        while temp > 0:
            digit = temp % 10
            digit_sum = digit_sum + digit
            temp = temp // 10

        print("Digit Sum:", digit_sum)

    case 4:
        reverse = 0
        temp = num

        while temp > 0:
            digit = temp % 10
            reverse = reverse * 10 + digit
            temp = temp // 10

        if num == reverse:
            print("Palindrome")
        else:
            print("Not Palindrome")

    case _:
        print("Invalid choice")
```

```python
# 35. Create a menu-driven conversion program for binary, decimal, and hexadecimal values.

print("1. Binary to Decimal")
print("2. Decimal to Binary")
print("3. Binary to Hexadecimal")
print("4. Hexadecimal to Binary")
print("5. Decimal to Hexadecimal")
print("6. Hexadecimal to Decimal")

choice = int(input("Enter your choice (1-6): "))

match choice:
    case 1:
        binary = input("Enter binary value: ")
        decimal = int(binary, 2)
        print("Decimal:", decimal)

    case 2:
        decimal = int(input("Enter decimal value: "))
        binary = bin(decimal)[2:]
        print("Binary:", binary)

    case 3:
        binary = input("Enter binary value: ")
        hexadecimal = hex(int(binary, 2))[2:]
        print("Hexadecimal:", hexadecimal)

    case 4:
        hexadecimal = input("Enter hexadecimal value: ")
        binary = bin(int(hexadecimal, 16))[2:]
        print("Binary:", binary)

    case 5:
        decimal = int(input("Enter decimal value: "))
        hexadecimal = hex(decimal)[2:]
        print("Hexadecimal:", hexadecimal)

    case 6:
        hexadecimal = input("Enter hexadecimal value: ")
        decimal = int(hexadecimal, 16)
        print("Decimal:", decimal)

    case _:
        print("Invalid choice")
```
