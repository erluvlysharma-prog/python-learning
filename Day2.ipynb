```python
# 12. Check whether three sides can form a valid triangle.
a = int(input("Enter the first side: "))
b = int(input("Enter the second side: "))
c = int(input("Enter the third side: "))
if a + b > c and b + c > a and c + a > b:
  print("Three sides can form a valid triangle")
else:
  print("Three sides can not form a valid triangle")
```

```python
# 13. Determine whether a triangle is equilateral, isosceles, or scalene.
a = int(input("Enter the first side: "))
b = int(input("Enter the second side: "))
c = int(input("Enter the third side: "))
if a == b == c:
  print("This triangle is equilateral.")
elif a == b and b == c and c == a:
  print("This triangle is isosceles.")
else:
  print("This triangle is scalene.")
```

```python
# 14. Determine whether a triangle is acute, right, or obtuse using its sides.
a = int(input("Enter the first side: "))
b = int(input("Enter the second side: "))
c = int(input("Enter the third side: "))
if a**2 + b**2 == c**2:
  print("This triangle is right angled.")
elif a**2 + b**2 < c**2:
  print("This triangle is acute angled.")
else:
  print("This triangle is obtuse angled.")
```

```python
# 16. Calculate electricity bill using slab-based conditions.
unit = int(input("Enter the unit"))
if unit <= 150:
  bill = unit* 5
elif 150 < unit <= 300:
  bill = unit * 5 + (unit - 150) * 7
else:
  bill = unit * 5 + unit * 7 + (unit - 150) * 9
print("Your electricity bill : ",bill)
```
